```python
## helper consts
years = list(range(2011, 2027))

normalizedColumns = [
    "fund_type",
    "fund_description",
    "department_description",
    "appropriation_authority_description",
    "appropriation_account_description",
]

replacements = [
    ["'S", "'s"],
    [" Of ", " of "],
    [" On ", " on "],
    [" The ", " the "],
    [" And ", " and "],
    [" For ", " for "],
    [" With ", " with "],
]

## helper functions
def getRowAmount(row):
    if not pd.isna(row["_ordinance_amount_"]):
        return row["_ordinance_amount_"]
    elif not pd.isna(row["appropriation_ordinance"]):
        return row["appropriation_ordinance"]
    else:
        return row["amount"]

def fmtCode(code, length, withInt=False):
    if isinstance(code, float) and math.isnan(code):
        return "NA"
    if withInt:
        if isinstance(code, str) and not code.isdigit():
            return code
        return "0" * (length - len(str(int(code)))) + str(int(code))
    return "0" * (length - len(str(code))) + str(code)

def getValues(df, x):
    subset = df[
        (df["department_number"] == x["department_number"])
        & (df["fund_code"] == x["fund_code"])
        & (df["appropriation_account"] == x["appropriation_account"])
        & (df["appropriation_authority"] == x["appropriation_authority"])
    ]
    values = subset[["amount"]].T
    values.columns = subset["year"].tolist()
    return values.T["amount"]

def cleanTitle(row):
    row = row.title()
    for replacement in replacements:
        row = row.replace(*replacement)
    return row

def makeSeries(ls, columns, titleCase=False):
    df = pd.DataFrame(ls).sort_values(by=[0])
    df.columns = columns
    df = df.set_index(columns[0])
    series = df[columns[1]]

    if titleCase:
        series = series.apply(cleanTitle)
    return series

def splitSeries(ls, columns, titleCase=False):
    return makeSeries(list(map(lambda x: x.split(" - ", 1), ls)), columns, titleCase)

def makeCSV(dfOrSeries, name):
    dfOrSeries.to_csv(f"./python/labeled/{name}.csv")
    dfOrSeries.to_csv(f"./page/data/{name}.csv")
```

```python
df = pd.DataFrame()

for year in years:
    dfYear = pd.read_csv(f"../data/{year}-ordinance.csv")
    dfYear["year"] = year
    df = pd.concat([df, dfYear], ignore_index=True)

# clean rows and columns
df["amount"] = df.apply(getRowAmount, axis=1)
df["fund_code"] = df["fund_code"].apply(lambda x: fmtCode(x, 4))
df["department_number"] = df["department_number"].apply(lambda x: fmtCode(x, 2))
df["appropriation_account"] = df["appropriation_account"].apply(lambda x: fmtCode(x, 4))
df["appropriation_authority"] = df["appropriation_authority"].apply(lambda x: fmtCode(x, 4, True))

for colName in normalizedColumns:
    df[colName] = df[colName].str.upper()

df.head()
```

```python
categoryMatches = []
for key, value in categories.items():
    for code in value:
        categoryMatches.append([code, key])

categorySeries = makeSeries(categoryMatches, ["deptNum", "category"])
deptNameSeries = splitSeries(depts, ["deptNum", "deptName"], True)

deptsDF = pd.concat([deptNameSeries, categorySeries], axis=1)
deptsDF.head()
```

```python
accountsSeries = splitSeries(accounts, ["accountNum", "accountName"])
fundsSeries = splitSeries(fundsCodes, ["fundNum", "fundName"])
fundTypes = (
    df[["fund_type", "fund_code"]]
    .value_counts()
    .to_frame()
    .drop(["count"], axis=1)
    .reset_index()
    .sort_values(by=["fund_code"])
    .set_index("fund_code")
)

fundsDF = pd.concat([fundsSeries, fundTypes], axis=1)
fundsDF
```

```python
cleaned = (
    df[
        [
            "department_number",
            "fund_code",
            "appropriation_account",
            "appropriation_authority",
        ]
    ]
    .value_counts()
    .to_frame()
    .drop(["count"], axis=1)
    .reset_index()
    .sort_values(
        by=[
            "department_number",
            "fund_code",
            "appropriation_account",
            "appropriation_authority",
        ]
    )
)

amounts = cleaned.merge(
    cleaned.apply(lambda row: getValues(df, row), axis=1, result_type="expand").fillna(0),
    left_index=True,
    right_index=True,
)

amounts.head()
```
